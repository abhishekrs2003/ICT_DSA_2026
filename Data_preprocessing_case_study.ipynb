###Outlier handling

```python
house_price_df.describe()
```

```python
plt.figure(figsize=(10,40))
for i, col in enumerate(num_cols,1):
  plt.subplot(9,2,i)
  house_price_df[col].plot(kind = 'box')
  plt.title(col)
  plt.grid()

plt.tight_layout()
plt.show()
```

```python
Q1 = house_price_df[num_cols].quantile(0.25)
Q3 = house_price_df[num_cols].quantile(0.75)
IQR = Q3-Q1


upper_limit = Q3 + (1.5 * IQR)
lower_limit = Q1 - (1.5 * IQR)

outliers = ((house_price_df[num_cols] > upper_limit) | (house_price_df[num_cols] < lower_limit)).any(axis=1)

outliers_df = house_price_df[outliers]

outlier_percentage = len(outliers_df) * 100/ len(house_price_df)

print("percentage fo outliers in the dataframe :", outlier_percentage)
```

```python
outliers_count = ((house_price_df[num_cols] > upper_limit) | (house_price_df[num_cols] < lower_limit)).mean()*100
```

```python
outliers_count
```

```python
clip_cols= ['Lot Area (in Sqft)','Lot Area after Renovation (in Sqft)','Overall Grade', 'Renovated Year']
```

```python
house_price_df[clip_cols] = house_price_df[clip_cols].clip(lower_limit, upper_limit, axis=1) # cliping outlier which have percentage greater than 3
```

```python
house_price_df = house_price_df[~outliers] # removing rest of the outliers
```

```python
plt.figure(figsize=(10,40))
for i, col in enumerate(num_cols,1):
  plt.subplot(9,2,i)
  house_price_df[col].plot(kind = 'box')
  plt.title(col)
  plt.grid()

plt.tight_layout()
plt.show()
```
